### import libraries

```python
import pandas as pd
import os
```

### Base path to src folder

```python
leases_file = os.path.join("..","src","leases.csv")
locations_file = os.path.join("..","src","locations.csv")
properties_file = os.path.join("..","src","properties.csv")
tenants_file = os.path.join("..","src","tenants.csv")
units_file = os.path.join("..","src","units.csv")
```

```python
print(leases_file)
```

#### Read CSVs into Pandas Dataframes

```python
leases = pd.read_csv(leases_file)
locations = pd.read_csv(locations_file)
properties = pd.read_csv(properties_file)
tenants = pd.read_csv(tenants_file)
units = pd.read_csv(units_file)
```

```python
dfs = [leases,locations,properties,tenants,units]
names = ["leases","locations","properties","tenanats","units"]
```

```python
for i,df in enumerate(dfs):
    print(f"Dafaframe for {names[i]}")
    print(df)
    print("\n")
```
